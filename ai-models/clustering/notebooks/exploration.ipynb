# Geo-Spatial Grievance Exploration

This notebook visualizes the DBSCAN outputs across city coordinates.

```python
import os
import sys
import pandas as pd
import numpy as np
import folium
from folium.plugins import HeatMap

# Add src to path
sys.path.append(os.path.abspath('../src'))
from dbscan import geo_clustering
```

```python
# Mock Grievance Data
grievances = [
    {'id': 1, 'lat': 28.6139, 'lng': 77.2090, 'priority': 'HIGH'},
    {'id': 2, 'lat': 28.6140, 'lng': 77.2088, 'priority': 'CRITICAL'},
    {'id': 3, 'lat': 28.6138, 'lng': 77.2092, 'priority': 'MODERATE'},
    {'id': 4, 'lat': 28.6141, 'lng': 77.2091, 'priority': 'HIGH'},
    {'id': 5, 'lat': 28.6137, 'lng': 77.2089, 'priority': 'MINOR'},
    
    # Cluster 2 (further away)
    {'id': 6, 'lat': 28.5355, 'lng': 77.2711, 'priority': 'CRITICAL'},
    {'id': 7, 'lat': 28.5356, 'lng': 77.2710, 'priority': 'CRITICAL'},
    {'id': 8, 'lat': 28.5354, 'lng': 77.2712, 'priority': 'CRITICAL'},
]

clusters = geo_clustering.detect_clusters(grievances)
print(f"Found {len(clusters)} clusters.")
clusters
```

```python
# Base Map
m = folium.Map(location=[28.6139, 77.2090], zoom_start=11)

# Add Heatmap layer
heat_data = [[g['lat'], g['lng']] for g in grievances]
HeatMap(heat_data).add_to(m)

# Add Cluster Markers
for c in clusters:
    folium.CircleMarker(
        location=[c['centroid']['lat'], c['centroid']['lng']],
        radius=c['crisis_score'] * 2,
        popup=f"Cluster {c['cluster_id']} (Score: {c['crisis_score']})",
        color="red" if c['crisis_score'] > 5 else "orange",
        fill=True,
    ).add_to(m)

m
```
